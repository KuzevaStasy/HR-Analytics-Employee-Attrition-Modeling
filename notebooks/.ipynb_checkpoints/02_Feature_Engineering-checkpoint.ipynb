# HR Analytics: Feature Engineering

**Goal:**  
Transform raw HR data into meaningful, model-ready features
that capture employee tenure, compensation context, engagement,
and potential attrition risk.

This notebook builds on insights from the EDA phase.

## 1. Data Loading & Setup

```python
import pandas as pd
import numpy as np

from datetime import datetime

pd.set_option("display.max_columns", None)
```

## 2. Load Dataset

```python
df = pd.read_csv("../data/raw/HRDataset_v14.csv")
df.head()
```

## 3. Target Variable: Attrition

```python
df["Attrition"] = np.where(df["EmploymentStatus"] == "Active", 0, 1)
df["Attrition"].value_counts(normalize=True)
```

**Explanation:**  
Attrition is defined as any form of employee termination, regardless of reason.
The target is encoded as a binary variable to support classification models
that estimate the probability of an employee leaving the company.

## 4. Date Processing & Tenure

```python
date_cols = [
    "DateofHire",
    "DateofTermination",
    "LastPerformanceReview_Date"
]

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors="coerce")
```

```python
# Reference date
today = pd.to_datetime("today")

# Create effective end date:
# - DateofTermination for terminated employees
# - today for active employees
df["EndDate"] = df["DateofTermination"].fillna(today)

# Tenure in years
df["TenureYears"] = (
    (df["EndDate"] - df["DateofHire"])
    .dt.days / 365
)
```

**Explanation:**  
Employee tenure represents the length of time an employee has been with the company.
For terminated employees, tenure ends at the termination date.
For active employees, tenure is calculated up to the current date.

This approach avoids data leakage and ensures that tenure reflects the employee's
actual employment lifecycle at the time of analysis.

## 5. Tenure Buckets

```python
df["TenureGroup"] = pd.cut(
    df["TenureYears"],
    bins=[0, 1, 3, 5, 10, 40],
    labels=["<1 year", "1–3 years", "3–5 years", "5–10 years", "10+ years"]
)
```

**Explanation:**  
Attrition risk is often non-linear across an employee’s tenure.
Early-stage employees typically have higher churn risk, while longer-tenured
employees tend to be more stable.

Tenure buckets allow models to capture these non-linear patterns
that may not be fully represented by a continuous tenure variable.

## 6. Relative Salary Features

```python
df["DeptAvgSalary"] = df.groupby("Department")["Salary"].transform("mean")

df["RelativeSalary"] = df["Salary"] / df["DeptAvgSalary"]
```

**Explanation:**  
Relative salary measures an employee’s compensation compared to the average salary
within their department.

This feature captures internal pay fairness and perceived equity, which can be
more influential for attrition than absolute salary levels alone.

## 7. Engagement & Satisfaction Risk Flags

```python
df["LowEngagementFlag"] = np.where(df["EngagementSurvey"] < 3.5, 1, 0)
df["LowSatisfactionFlag"] = np.where(df["EmpSatisfaction"] <= 3, 1, 0)
```

**Explanation:**  
Binary risk flags make it easier for tree-based models
to identify high-risk employee segments.

## 8. Attendance & Behavior Features

```python
df["HighAbsenceFlag"] = np.where(df["Absences"] > df["Absences"].median(), 1, 0)
df["LateRecentlyFlag"] = np.where(df["DaysLateLast30"] > 0, 1, 0)
```

**Explanation:**  
Attendance-related behaviors, such as frequent absences or recent lateness,
may reflect disengagement or personal challenges.

While these signals are weaker predictors on their own, they can complement
engagement and satisfaction features when combined in a predictive model.

## 9. Performance Encoding

```python
performance_map = {
    "PIP": 1,
    "Needs Improvement": 2,
    "Fully Meets": 3,
    "Exceeds": 4
}

df["PerformanceScoreNum"] = df["PerformanceScore"].map(performance_map)
```

**Explanation:**  
Performance ratings follow a clear ordinal structure.
Encoding them numerically preserves their ranking while allowing models
to learn how performance level relates to attrition risk.

## 10. Feature Selection for Modeling

```python
model_features = [
    "TenureYears",
    "RelativeSalary",
    "EngagementSurvey",
    "EmpSatisfaction",
    "SpecialProjectsCount",
    "DaysLateLast30",
    "Absences",
    "PerformanceScoreNum",
    "LowEngagementFlag",
    "LowSatisfactionFlag",
    "HighAbsenceFlag",
    "LateRecentlyFlag"
]

X = df[model_features]
y = df["Attrition"]
```

**Explanation:**  
The selected features represent a combination of tenure, compensation,
engagement, performance, and behavioral indicators.

This balanced feature set is designed to capture both structural and
attitudinal drivers of employee attrition.

## 11. Save Processed Dataset

```python
df_model = pd.concat([X, y], axis=1)
df_model.to_csv("../data/processed/hr_model_ready.csv", index=False)
```

**Outcome:**  
The resulting dataset is fully numeric, clean, and suitable for use
in machine learning pipelines focused on employee attrition prediction.

## Feature Engineering Summary

This notebook transformed raw HR data into model-ready features by:

- Representing employee lifecycle through tenure-based features
- Capturing internal compensation fairness using relative salary
- Highlighting disengagement and dissatisfaction through risk indicators
- Encoding performance and attendance behavior
- Producing a clean dataset suitable for explainable and predictive modeling

These features provide a strong foundation for building attrition prediction models.
